```python
import pandas as pd

# Read the CSV file.
# Adjust the separator if necessary. If your file is tab-delimited, use sep='\t'
# For a comma-separated file, you can omit the sep argument or use sep=','
df = pd.read_csv(r"C:\Users\brand\Documents\real-estate-momentum\data\raw\Metro_zhvi_uc_sfr_tier_0.33_0.67_sm_sa_month.csv")  # change the file name and delimiter as needed

# Check the first few rows to understand its structure
print("Original Data:")
print(df.head())

# Identify the columns that hold the date values.
# In your CSV sample, the date columns start at the 6th column.
date_columns = df.columns[5:]  # This assumes columns 0-4 are metadata (RegionID, SizeRank, RegionName, RegionType, StateName)
```

```python
df
```

```python
# Use pd.melt to transform the wide-format data into long-format.
df_long = pd.melt(
    df,
    id_vars=["RegionName", "SizeRank", "StateName"],  # Keeping additional columns
    value_vars=date_columns,  # The columns that represent dates
    var_name="Date",          # New column to hold the date values
    value_name="Price"        # New column to hold the price values
)

# Rename the 'RegionName' column to 'Metro' to match your desired output.
df_long = df_long.rename(columns={"RegionName": "Metro"})

# Convert the 'Date' column from string to datetime.
df_long["Date"] = pd.to_datetime(df_long["Date"])

# Display the transformed DataFrame
print("\nTransformed Data (Metro, Date, Price, SizeRank, StateName):")
print(df_long.head())

# Optionally, save the transformed data to a new CSV file.
df_long.to_csv("../transformed_price_data.csv", index=False)
```
